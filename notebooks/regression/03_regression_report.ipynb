# Predictive Modeling Report

This notebook summarizes the results of the predictive modeling efforts for NYC taxi demand at the **zone × hour** level. The objective was to quantify the influence of temporal and trip-related features on demand and evaluate model performance.

---

## **1. Modeling Objective**

- Predict **taxi demand** (`demand`) using:
  - Temporal features: `hour`, `day_of_week`, `is_weekend`
  - Trip characteristics: `avg_fare`, `avg_distance`
- Compare **StatsModels OLS regression** with **sklearn LinearRegression**
- Provide insights for operational planning and potential business use cases.

---

## **2. Data Overview**

| Feature        | Description |
|----------------|------------|
| hour           | Hour of the day (0–23) |
| day_of_week    | Day of the week (1=Monday … 7=Sunday) |
| is_weekend     | Binary weekend indicator (1=yes, 0=no) |
| avg_fare       | Average fare of trips in the zone-hour |
| avg_distance   | Average trip distance in the zone-hour |
| demand         | Number of completed rides (target) |

Data is aggregated at the **zone × hour** level from January–March 2019.

---

## **3. Model Summary**

### **3.1 StatsModels OLS Regression**

- **R²:** 0.085 → ~8.5% of variance in demand explained  
- **Key coefficients:**

| Feature        | Coefficient |
|----------------|------------|
| const          | 68.18      |
| hour           | 3.60       |
| day_of_week    | 2.23       |
| is_weekend     | -14.28     |
| avg_fare       | -0.93      |
| avg_distance   | -4.37      |

**Insights:**

- Demand increases during the day (`hour`) and slightly later in the week (`day_of_week`).  
- Weekend demand is substantially lower (`is_weekend`).  
- Higher fare or longer trips reduce ride frequency (`avg_fare`, `avg_distance`).  
- Residuals are right-skewed with heavy tails, indicating that extreme high-demand periods are not fully captured by the linear model.  
- Strong positive autocorrelation is present, as expected in time-series data.

---

### **3.2 Sklearn Linear Regression**

- **R² on test set:** 0.087  
- **MSE on test set:** 18,624  

| Feature        | Coefficient |
|----------------|------------|
| hour           | 3.57       |
| day_of_week    | 2.18       |
| is_weekend     | -14.06     |
| avg_fare       | -0.93      |
| avg_distance   | -4.33      |

**Insights:**

- Coefficient signs and magnitudes are consistent with the OLS model.  
- Slightly higher R² on the test set confirms reproducibility of results with standard ML workflow.  
- Model predicts general demand trends but misses extreme peaks and localized surges.

---

## **4. Model Interpretation**

- **Hour effect:** +3.57 per additional hour → demand rises through the day.  
- **Weekend effect:** -14 → lower activity compared to weekdays.  
- **Fare effect:** -0.93 → higher fare slightly decreases demand.  
- **Distance effect:** -4.33 → longer trips occur less frequently.  

**Scenario Example:**

- Increasing time by 5 hours → ~18 additional rides  
- Increasing trip distance by 2 units → ~9 fewer rides  

---

## **5. Limitations**

- Linear relationships may not capture real-world peaks and troughs in demand.  
- Important predictors (location, weather, events) are missing.  
- Residual autocorrelation violates OLS assumptions.  
- Non-normality of residuals indicates that extreme demand spikes are underrepresented.

---

## **6. Business Implications**

- Models provide **high-level insights** on demand drivers but are insufficient for precise operational forecasting.  
- **Key operational takeaways:**
  - **Driver allocation:** Increase supply during peak hours; reduce during off-peak periods.  
  - **Dynamic pricing:** Higher fares reduce frequency, which could inform surge pricing strategies.  
  - **Zone-specific planning:** Extreme demand events are not fully captured, highlighting the need for more granular features.

---

## **7. Conclusion**

- Linear regression identifies **meaningful relationships** between temporal/trip features and demand.  
- The limited R² suggests **missing drivers**, but model insights are actionable for operational decision-making.
